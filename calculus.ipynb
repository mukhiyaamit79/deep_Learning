```python
x = sp.symbols('x')

P = -2*x**2 + 80*x - 300

# First derivative
P_prime = sp.diff(P, x)
print("First derivative:", P_prime)
```

```python
# Find critical points
critical_points = sp.solve(P_prime, x)
print("Critical points:", critical_points)
```

```python
# Second derivative
P_second = sp.diff(P_prime, x)
print("Second derivative:", P_second)
```

```python
# Maximum profit
for point in critical_points:
    print("Profit at critical point:", P.subs(x, point))
```
